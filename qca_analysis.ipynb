```python
import numpy as np
```

```python
def qca_gibbs_binary(na, nb, naa, nab, nbb, total_pairs):
    total = na + nb
    xa, xb = na / total, nb / total
    naa, nab, nbb, total_pairs = [np.asarray(v, dtype=float)for v in (naa, nab, nbb, total_pairs)]
    Xaa, Xab, Xbb = naa / total_pairs, nab / total_pairs, nbb / total_pairs
    # free energy (kT)
    dg_ab = -np.log(Xab**2 / (4.0 * Xaa * Xbb))
    # enthalpic term
    deltaH = 0.5 * (nab * dg_ab) / total
    # ideal configurational entropy
    S_point = -(na * np.log(xa) + nb * np.log(xb)) / total
    # correlation entropy
    S_pair = -(naa * np.log(Xaa / xa**2)+ nbb * np.log(Xbb / xb**2)+ nab * np.log(Xab / (2.0 * xa * xb))) / total
    return {"dg_ab": dg_ab, "deltaH": deltaH, "S_point": S_point,"S_pair": S_pair, "G_point": deltaH - S_point,"G_mix": deltaH - S_point - S_pair}
```

```python
path=f'/home/shivam/data/SG-4_line/4_unsaturation_diff/DRPC_DIPC/analysis/entropy/upper'
gibbs_file=np.loadtxt(path+'/interaction_matrix.dat',comments='#',unpack=True)
naa=gibbs_file[1]
nab=gibbs_file[2]
nbb=gibbs_file[3]
total_pairs=gibbs_file[4]
na=1065; nb=455
qca_gibbs_binary=qca_gibbs_binary(na, nb, naa, nab, nbb, total_pairs)
gibbs_energy_mixing=np.mean(qca_gibbs_binary["G_mix"])
```
